```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# 1. Create some sample data
np.random.seed(42)
data = {
    'Category': np.random.choice(['A', 'B', 'C', 'D'], size=100),
    'Value': np.random.randint(1, 100, size=100)
}
df = pd.DataFrame(data)

print("Original DataFrame head:")
print(df.head())
print("\n")

# 2. Basic Data Analysis
# Calculate descriptive statistics
print("Descriptive Statistics for 'Value':")
print(df['Value'].describe())
print("\n")

# Group by 'Category' and calculate mean of 'Value'
category_mean = df.groupby('Category')['Value'].mean().reset_index()
print("Mean 'Value' by Category:")
print(category_mean)
print("\n")

# 3. Data Visualization
# Create a bar plot of mean 'Value' by 'Category'
plt.figure(figsize=(8, 6))
plt.bar(category_mean['Category'], category_mean['Value'], color='skyblue')
plt.xlabel('Category')
plt.ylabel('Mean Value')
plt.title('Mean Value by Category')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()

print("Basic data analysis complete. Dataframe, statistics, and a plot have been generated.")
```
